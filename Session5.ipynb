Create and retrieve a product record 

```python
product = {'id': 101, 'name': 'Notebook', 'price': 125.0}
print("Initial values:", product)

product_id = product['id']
product_name = product['name']
product_price = product['price']

print("Product ID:", product_id)
print("Product Name:", product_name)
print("Product Price:", product_price)
```

Add and update dictionary entries 

```python
product = {'name': 'Pen', 'price': 20.0}
stock = 50
revised_price = 22.0
print("Original:", product)

product['stock'] = stock
product['price'] = revised_price
print("Stock added:", product)
print("Final product record:", product)
```

Use get() for a possibly missing key 

```python
employee = {'id': 'E101', 'name': 'Ravi'}
requested_key = 'department'

print("Original employee:", employee)
print("Requested key:", requested_key)

department = employee.get(requested_key, 'Not available')
print("Value retrieved:", department)
print("Final result:", department)
```

Remove an entry with pop()

```python
cart = {'Pen': 3, 'Notebook': 2, 'Marker': 1}
item_to_remove = 'Marker'

print("Original:", cart)
print("Item to remove:", item_to_remove)

removed_quantity = cart.pop(item_to_remove)
print("Removed quantity:", removed_quantity)
print("Remaining cart items:", cart)
```

Retrieve values from a nested dictionary 

```python
student = {
    'name': 'asha',
    'marks': {
        'Python': 82,
        'Statistics': 76
    }
}

print("Original:", student)

python_marks = student['marks']['Python']
statistics_marks = student['marks']['Statistics']

print("Python marks:", python_marks)
print("Statistics marks:", statistics_marks)
```

Merge two dictionaries with the union operator 

```python
defaults = {'theme': 'light', 'page_size': 20}
user = {'theme': 'dark'}
print("Default:", defaults)
print("User settings:", user)

settings = defaults | user
print("Merged settings:", settings)
```

Inspect keys, values, and items views 

```python
inventory = {'Pen': 40, 'Notebook': 25, 'Marker': 18}
print("Original:", inventory)

keys_view = inventory.keys()
values_view = inventory.values()
items_view = inventory.items()

print("Keys:", keys_view)
print("Values:", values_view)
print("Items:", items_view)
```

Compare a dictionary alias with a copy 

```python
original = {'theme': 'light', 'page_size': 20}
print("Original:", original)

alias = original
copied = original.copy()

print("Alias before update:", alias)
print("Copied before update:", copied)

original['theme'] = 'dark'

print("After updating:")
print("Alias:", alias)
print("Original:", original)
print("Copied:", copied)
```
