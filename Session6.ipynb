Store quotient and remainder in a tuple 

```python
dividend = 137
divisor = 60
print("Dividend:", dividend)
print("Divisor:", divisor)

quotient = dividend // divisor
remainder = dividend % divisor

print("Quotient:", quotient)
print("Remainder:", remainder)

result= (quotient, remainder)

print("Result Tuple:", result)

unpacked_quotient, unpacked_remainder = result

print("Unpacked quotient:", unpacked_quotient)
print("Unpacked remainder:", unpacked_remainder)
```

Write and read a complete text file 

```python
from pathlib import Path

path_text = 'North\nSouth\nEast\n'
print("Starting values:", path_text)

file_path = Path("directions_temp.txt")
file_path.write_text(path_text, encoding="utf-8")
print("Text written to:", file_path)

read_text = file_path.read_text(encoding="utf-8")
print("Text read from file:", read_text)

file_path.unlink()
print("Temporary file deleted.")
print("Final result:", read_text)
```

Copy text after converting it to uppercase 

```python
from pathlib import Path

source_text = 'Python files are readable.'
print("Source text:", source_text)

source_path = Path("source_temp.txt")
uppercase_path = Path("uppercase_temp.txt")
source_path.write_text(source_text, encoding="utf-8")
read_text = source_path.read_text(encoding="utf-8")
print("Text read from first file:", read_text)
uppercase_text = read_text.upper()
print("Uppercase intermediate value:", uppercase_text)
uppercase_path.write_text(uppercase_text, encoding="utf-8")
verified_text = uppercase_path.read_text(encoding="utf-8")
print("Text read from second file:", verified_text)
source_path.unlink()
uppercase_path.unlink()

print("Final result:", verified_text)
```

Read three known lines explicitly

```python
from pathlib import Path

line_1 = 'North'
line_2 = 'South'
line_3 = 'East'

print("Line 1:", line_1)
print("Line 2:", line_2)
print("Line 3:", line_3)

file_path = Path("three_lines_temp.txt")
file_path.write_text(
    line_1 + '\n' + line_2 + '\n' + line_3 + '\n',
    encoding="utf-8"
)

lines = file_path.read_text(encoding="utf-8").splitlines()
print("\nAll lines read:", lines)

first_line = lines[0]
second_line = lines[1]
third_line = lines[2]
print("First line:", first_line)
print("Second line:", second_line)
print("Third line:", third_line)
file_path.unlink()

print("Final Results:", first_line, second_line, third_line)
```

Write CSV-style records without a loop 

```python
from pathlib import Path

header = 'name,score'
row_1 = 'Asha,82'
row_2 = 'Ravi,76'

print("Header:", header)
print("Row 1:", row_1)
print("Row 2:", row_2)

csv_text = header + '\n' + row_1 +'\n' + row_2 + '\n'

print("\nCSv text before writing:", csv_text)

file_path = Path("records_temp.csv")
file_path.write_text(csv_text, encoding="utf-8")
read_text = file_path.read_text(encoding="utf-8")
print("Text read nack:", read_text)

file_path.unlink()
print("Final result:", read_text)
```

Encode and decode Unicode text with UTF-8 

```python
message = 'नमस्ते Python'
print("Message:", message)
encoded_message = message.encode('utf-8')
print("Encoded message:", encoded_message)

decoded_message = encoded_message.decode('utf-8')
print("Decoded message:", decoded_message)
is_equal = decoded_message == message
print("Decoded message is equal to the original message:", is_equal)
```
