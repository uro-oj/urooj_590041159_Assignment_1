Create, index, and slice a product list 

```python
products = ['Pen', 'Notebook', 'Marker', 'Folder', 'Stapler']
print("The list:", products)

first_product = products[0]
last_product = products[-1]
middle_product = products[1:4]
reversed_products = products[::-1]

print("First prodcut:", first_product)
print("Last product:", last_product)
print("Middle products:", middle_product)
print("Reversed products:", reversed_products)
```

Compare append() with extend()

```python
items = ['Pen', 'Notebook']
new_items = ['Markers', 'Folder']

print("Main list:", items)
print("New items:", new_items)

append_result = items.copy()
append_result.append(new_items)

extend_result = items.copy()
extend_result.extend(new_items)

print("Appended:", append_result)
print("Extended:", extend_result)
print("Comparison:", append_result == extend_result)
```

Use insert(), remove(), and pop() 

```python
cities = ['Delhi', 'Mumbai', 'Pune', 'Jaipur']
removed_city = 'Pune'

print("Cities:", cities)

cities.insert(1, 'Chennai')
print("After update:", cities)
print("Before Removal:", cities)

cities.remove(removed_city)
print("After removal:", cities)

final_city = cities.pop()
print("After pop:", cities)
print("Final city:", final_city)
```

Compare sort() with sorted() 

```python
scores = [72, 91, 68, 85, 77]
print(scores)

sorted_scores = sorted(scores)
print("Scores after sorted():", sorted_scores)

scores.sort(reverse=True)
print("Scores after sort(reverse=True):", scores)

print("Comparison:", "sorted() result:", sorted_scores, "sort() result:", scores)
```

Retrieve values from a nested list 

```python
matrix = [[10, 20], [30, 40], [50, 60]]
print("Initial list:", matrix)

value1 = matrix[0][0]
value2 = matrix[1][1]
value3 = matrix[2][1]

print("Selected values:", value1, value2, value3)
result = [value1, value2, value3]
print("Final result:", result)
```

Demonstrate list aliasing 

```python
original = [10, 20, 30]
print("Original list:", original)

alias = original
print("After alias = original:")
print("Alias:", alias)
print("Original:", original)

alias.append(40)
print("After Appending 40:")
print("Alias:", alias)
print("Original:", original)
```

Make an explicit independent nested copy 

```python
original = [[1], [2]]
print("Original list:", original)

independent = [original[0].copy(), original[1].copy()]
print("Independent copy:", independent)

original[0].append(99)
print("After mutating:")
print("Original:", original)
print("Independent:", independent)
```

Combine and repeat lists 

```python
first_list = ['A', 'B']
second_list = ['C', 'D']
repeat_count=2 

print("Starting first list:", first_list)
print("Starting second list:", second_list)
print("Starting repeat count:", repeat_count)

combined = first_list + second_list
print("Combined:", combined)

repeated = combined * repeat_count
print("repeated:", repeated)
```
