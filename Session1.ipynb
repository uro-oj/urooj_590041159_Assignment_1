Convert Celsius to Fahrenheit
Calculate fahrenheit from celcius by using the standard formula and display both temperatures with units.

```python
celsius = 32.0
fahrenheit = ( celsius * 9 / 5 ) + 32
print("Temperature in Celsius:", celsius, "C")
print("Temperature in Fahrenheit:", fahrenheit, "F")

celsius = 0.0
fahrenheit = (celsius * 9 / 5) + 32
print("Temperature in Celsius:", celsius, "C")
print("Temperature in fahrenheit:", fahrenheit, "F")
```

Convert total seconds to hours, minutes, and seconds 

```python
total_seconds = 7384

print("Starting Seconds", total_seconds)

hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print("Hours:", hours)
print("Seconds:", remaining_seconds)
print("Minutes:", minutes)
print("Remaining seconds after minutes:", seconds)
print("Final result:", hours, "hours", minutes, "minutes", seconds, "seconds")

total_seconds = 59
print("Starting seconds:", total_seconds)

hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print("Hours:", hours)
print("Seconds:", remaining_seconds)
print("Minutes:", minutes)
print("Remaining seconds:", seconds)
print("Final result", hours, "hours", minutes, "minutes", seconds, "seconds")
```

Calculate a bill after percentage discount

```python
price = 2400.0
discount_rate = 0.15

print("Starting values:", price, discount_rate)

discount_amount = price * discount_rate
final_price = price - discount_amount

print("Discount amount:", discount_amount)
print("Final price:", final_price)
```
