Compute quotient and remainder 

```python
dividend = 137
divisor = 60

quotient = dividend // divisor
remainder = dividend % divisor

print("Quotient:", quotient)
print("Remainder:", remainder)
print("Verification:", dividend + quotient * divisor + remainder)
```

Round a measurement to two places 

```python
measurement =  18.3764
rounded_value = round(measurement, 2)
print("Rounded value:", rounded_value)
```

Convert rupees to paise exactly

```python
rupees = 125.75
paise = rupees * 100
paise = round(paise)

print("Starting value:", rupees)
print("After coversion:", rupees * 100, "paise")
print("Whole number result:", paise, "paise")
```

Calculate compound growth

```python
principle = 50000.0
annual_rate = 0.08
years = 2

growth_factor = 1 + annual_rate
final_amount = principle * growth_factor ** years
growth_amount = final_amount - principle

print ("Growth factor:", growth_factor)
print("Final Amount:", final_amount)
print("Growth amount:", growth_amount)
```

Find distance between two numeric points 

```python
point_a = 18.5
point_b = 42.75

difference = point_b - point_a
distance = abs(difference)

print("Point A:", point_a)
print("Point B:", point_b)
print("Difference:", difference)
print("Distance:", distance)
```

Test divisibility by 3 and 5 

```python
number = 45
divisible_by_3 = number % 3 == 0
divisible_by_5 = number % 5 == 0
print("Number:", number)
print("Divisible by 3:", divisible_by_3)
print("Divisible by 5:", divisible_by_5)
```

Convert bytes to KB and MB 

```python
byte_count = 5242880
kilobytes = byte_count / 1024
megabytes = kilobytes / 1024

print("Starting value:", byte_count, "bytes")
print("Kilobytes:", kilobytes, "KB")
print("Megabytes:", megabytes, "MB")
```

Evaluate a polynomial for x 

```python
x = 4
a = 2
b = 3
c = 5

squared_term = a * x ** 2
linear_term = b * x
polynomial_value = squared_term + linear_term + c

print("Squared terms:", squared_term)
print("Linear term:", linear_term)
print("Polynomial value:", polynomial_value)
```
