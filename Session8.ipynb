Classify a number as negative, zero, or positive

```python
number = -7
print("Number:", number)

if number < 0:
    classification = 'negative'
elif number == 0:
    classification = 'zero'
else:
    classification = 'positive'

print ("Classification:", classification)
```

Assign a letter grade with validated bounds 

```python
score = 84
print("Score:", score)

if 0 <= score <=100:
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = "F"
else: 
    grade = "Invalid"

print("Validated score:", score)
print("Grade assigned:", grade)
```

Test leap-year rules 

```python
year = 2028
print("Starting year:", year)

divisible_by_400 = year % 400 == 0
divisible_by_4 = year % 4 == 0
not_divisible_by_100 = year % 100 != 0

print("Divisible by 400:", divisible_by_400)
print("Divisible by 4:", divisible_by_4)
print("Not divisible by 100:", not_divisible_by_100)

is_leapyear = divisible_by_400 or(
    divisible_by_4 and not_divisible_by_100
)

print("Is leap year?", is_leapyear)
```

Validate username and password conditions 

```python
username = 'asha_91'
password = 'Data@123'

print("Username:", username)
print("Password:", password)

username_valid = len(username) >= 6
password_valid = len(password) >= 8
password_contains_at = '@' in password

print("Username length valid:", username_valid)
print("Password length valid:", password_valid)
print("Password contains @:", password_contains_at)

valid_credentials = (
    username_valid
    and password_valid
    and password_contains_at
)
print("Final validation:", valid_credentials)
```

Classify a triangle from side lengths 

```python
side_1 = 5
side_2 = 5
side_3 = 8

print("Sides:", side_1, side_2, side_3)

valid_triangle = (
    side_1 + side_2 > side_3
    and side_1 + side_3 > side_2
    and side_2 + side_3 > side_1

)

if not valid_triangle:
    triangle_type = 'invalid'
elif side_1 == side_2 and side_2 == side_3:
    triangle_type = 'equilateral'
elif side_1 == side_2 or side_1 == side_3 or side_2 == side_3:
    triangle_type = "isosceles"
else:
    triangle_type = "scalene"

print("Validity:", valid_triangle)
print("Triangle classification:", triangle_type)
```

Select a weekday with match and case 

```python
day_number = 3
print("Day number:", day_number)

match day_number:
    case 1:
        weekday = "Monday"
    case 2:
        weekday = "Tuesday"
    case 3:
        weekday = "Wednesday"
    case 4:
        weekday = "Thursday"
    case 5:
        weekday = "Friday"
    case 6:
        weekday = "Saturday"
    case 7:
        weekday = "Sunday"
    case _:
        weekday = "invalid"

print("Selected weekday:", weekday)
```
