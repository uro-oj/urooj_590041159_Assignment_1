Count vowels case-insensitively 

```python
text = 'Data Science'
lower_text = text.lower()

vowel_count = (
    lower_text.count('a')
    + lower_text.count('e')
    + lower_text.count('i')
    + lower_text.count('o')
    + lower_text.count('u')

)

print("Text:", text)
print("Lowercase text:", lower_text)
print("Vowel Count", vowel_count)
```

Reverse text without reversed

```python
text = 'Python'
reversed_text = text[::-1]

print("Original text:", text)
print("Reversed text:", reversed_text)
```

Test a normalized palindrome 

```python
text = '  Never Odd or Even '
stripped_text = text.strip()
lowercase_text = stripped_text.lower()
normalized_text = lowercase_text.replace(' ','')
reversed_text = normalized_text[::-1]
is_palindrome = normalized_text == reversed_text

print("Text:", text)
print("Stripped text:", stripped_text)
print("Lowercase text:", lowercase_text)
print("Normalized text:", normalized_text)
print("Reversed normalized text:", reversed_text)
print("Is palindrome:", is_palindrome)
```

Replace repeated whitespace 

```python
text ='Python  makes  text  readable' 
repeated_space = '   '
new_text = text.replace(repeated_space, ' ')

print("Text:", text)
print("Repeated value:", repr(repeated_space))
print("Final result:", new_text)
```

Create initials from a full name 

```python
full_name = 'Asha Mehta Singh'
name_parts = full_name.split()
initials = name_parts[0][0] + '.' + name_parts[1][0] + '.' + name_parts[2][0]

print("Full Name:", full_name)
print("Name parts:", name_parts)
print("Initials:", initials)
```

Count occurrences of a word 

```python
sentence = 'data helps data teams'
target = 'data'
occurance_count = sentence.count(target)

print("Sentence:", sentence)
print("Target word:", target)
print("Occurance no.:", occurance_count)
```

Format a receipt line with two decimals 

```python
item = 'Notebook' 
quantity = 3
unit_price = 125.5
line_total = quantity * unit_price
receipt_line = f"{item} | Qty: {quantity} | Unite Price: {unit_price:.2f} | Total:{line_total:.2f}"

print("Line total:", line_total)
print("Receipt:", receipt_line)
```
