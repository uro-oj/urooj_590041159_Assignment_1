Sum integers from 1 through n 

```python
n= 10
print("n:", n)

total = 0

for number in range(1, n + 1):
    total += number
    print("Adding", number, "-> running total:", total)

print("Final value:", total)
```

Print a multiplication table 

```python
number = 7

print("Number:", number)

for multiplier in range(1, 11):
    product = number * multiplier
    print("Table row:", number, "x", multiplier, "=", product)

print("Result:", product)
```

Compute factorial iteratively 

```python
number = 6
print("Number:", number)

factorial = 1
print("factorial =", factorial)

for value in range(1, number + 1):
    factorial = factorial * value
    print("After multiplying by", value, "-> factorial:", factorial)
print("Final_result:", factorial)
```

Count digits in an integer 

```python
number = 50842
print("Number:", number)
working_number = number
digit_count = 0
print("Initial working number;", working_number)

while working_number > 0:
    working_number = working_number // 10
    digit_count += 1
    print("After // 10:", working_number, "| Digit count:", digit_count)
final_result= digit_count

print("final result:", final_result)
```

Find the first matching list value 

```python
values = [12, 7, 18, 5, 21]
target = 18
print("Value:", values)
print("Target:", target)

found_value = None

for value in values:
    print("Checking values:", value)

    if value == target:
        found_value = value
        print("Match Found:", found_value)
        break
print("Final result:", found_value)
```

Read values until a sentinel 

```python
entries = ['12', '7', '5', 'done']
print("Entries:", entries)

total = 0

for entry in entries:
    if entry == 'done':
        break

    value = int(entry)
    total += value
    print("converted value:", value, "| running totals:", total)

final_result = total

print("Final result:", total)
```

Print a right triangle pattern 

```python
rows = 5
print("Rows:", rows)
pattern = ""

for row in range(1, rows+1):
    line = "*" * row
    pattern += line + "\n"
    print("Row", row, ":", line)

final_result = pattern
print("Final result:", final_result)
```

Test primality with a loop 

```python
number = 29
print("Number:", number)
is_prime = True

for divisor in range(2, number):
    divisible = number % divisor == 0
    print("Testing divisor:", divisor, "| Divisible:", divisible)

    if divisible:
        is_prime = False
        print("Divisor found:", divisor)
        break

print("Final Result:", is_prime)
```
