Use extended sequence unpacking 

```python
values = [10, 20, 30, 40, 50]
print("Starting values:", values)

first, *middle, last = values 

print("First:", first)
print("Middle:", middle)
print("Last:", last)
```

Demonstrate equality versus identity 

```python
first = [1, 2]
second = [ 1, 2]
print("First list:", first)
print("Second list:", second)

alias = first
print("Alias:", alias)

equality_result = first == second
identity_result = first is second
alias_identity = first is alias

print("first==second:", equality_result)
print("first is second:", identity_result)
print("first is alias:", alias_identity)
```

Swap three variables by unpacking 

```python
a = 'North'
b = 'South'
c = 'East'

print("a =", a)
print("b =", b)
print("c =", c)

a, b, c = b, c, a

print("Swapped:")
print("a =", a)
print("b =", b)
print("c =", c)
```

Avoid chained-assignment aliasing 

```python
first = []
second = []
print("First list:", first)
print("Second list:", second)

first.append(1)
print("First List:", first)
print("Second list:", second)
```

Update a list with augmented assignment 

```python
values = [10, 20]
extra_values =[30, 40]
print("Original list:", values)
print("Extra values:", extra_values)

values += extra_values
print("Renewed values:", values)
```

Capture a computed value with the walrus operator 

```python
numbers = [3, 8, 12]
print("List:", numbers)

largest = None
largest = max(numbers)
comparison = largest > 10

print("Largest:", largest)
print("Largest > 10:", comparison)
```

Unpack nested records 

```python
record = ('E101', ('Asha', 82))
print("Record:", record)

employee_id, (employee_name, employee_score) = record
print("Employee ID:", employee_id)
print("Employee Name:", employee_name)
print("Employee score:", employee_score)
```
